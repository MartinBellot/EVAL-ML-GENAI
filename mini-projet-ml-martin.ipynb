## Exercice 4 : Modélisation
### Question 4.1 : Remise à l'échelle
- Décrivez les distributions des différentes variables explicatives.
- Est-il nécessaire de mettre à l'échelle les données du dataset ? Justifiez votre réponse. Si c'est le cas faites-le.


````python

fig, axes = plt.subplots(3, 4, figsize=(16, 10))
axes = axes.flatten()

for i, col in enumerate(X.columns):
    sns.histplot(X[col], bins=40, kde=True, ax=axes[i], color='steelblue')

#set labels:
for i in range(12):
    axes[i].set_ylabel('Fréquence')
    axes[i].set_xlabel(X.columns[i])


plt.suptitle('Distribution des variables explicatives', fontsize=13, y=1.01)
plt.tight_layout()
plt.show()

#la fréquence représente le nombre d'observations pour chaque intervalle de valeurs de la variable.
````

````python

from sklearn.preprocessing import StandardScaler

# On exclut la colonne 'type' (déjà encodée 0/1) de la normalisation
cols_to_scale = [c for c in X.columns if c != 'type']

scaler = StandardScaler()
X_train_scaled = X_train.copy()
X_test_scaled  = X_test.copy()

X_train_scaled[cols_to_scale] = scaler.fit_transform(X_train[cols_to_scale])
X_test_scaled[cols_to_scale]  = scaler.transform(X_test[cols_to_scale])

print("Mise à l'échelle effectuée (StandardScaler).")
X_train_scaled.describe().round(2)
````

Les variables ont des **échelles très différentes** , donc on applique un **StandardScaler** (centrage-réduction : moyenne 0, écart-type 1) uniquement sur le subset d'entraînement, puis on applique la même transformation au subset de test. La colonne type est deja binaire (0 ou 1), elle est donc exclue de la normalisation.

### Question 4.2 : Création du modèle
- Choisissez un modèle sklearn pour l'entrainer et créer le votre. [Documentation](https://scikit-learn.org/stable/supervised_learning.html)
- Exemples:
>- Pour une classification, vous pouvez utiliser l'un de ces modèles: [RidgeClassifier](https://scikit-learn.org/stable/modules/linear_model.html#classification), [SVC](https://scikit-learn.org/stable/modules/svm.html#classification), [DecisionTreeClassifier](https://scikit-learn.org/stable/modules/tree.html#classification), [MLPClassifier](https://scikit-learn.org/stable/modules/neural_networks_supervised.html#classification), [RandomForestClassifier](https://scikit-learn.org/stable/modules/ensemble.html#classification), etc.
>- Pour une régression, vous pouvez utiliser l'un de ces modèles: [SVR](https://scikit-learn.org/stable/modules/svm.html#regression), [DecisionTreeRegressor](https://scikit-learn.org/stable/modules/tree.html#regression), [MLPRegressor](https://scikit-learn.org/stable/modules/neural_networks_supervised.html#regression), [RandomForestRegressor](https://scikit-learn.org/stable/modules/ensemble.html#regression), etc.

````python

from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train_scaled, y_train)
print("Modèle entraîné : RandomForestClassifier (n_estimators=100)")
````

### Question 4.3 : Evaluation du modèle
- Calculez la précision du score sur le subset d'entrainement puis sur le subset de test

````python

from sklearn.metrics import classification_report

train_score = model.score(X_train_scaled, y_train)
test_score  = model.score(X_test_scaled, y_test)

print(f"Score (accuracy) - Train : {train_score:.4f}")
print(f"Score (accuracy) - Test  : {test_score:.4f}")
print()
print("Rapport de classification (Test) :")
print(classification_report(y_test, model.predict(X_test_scaled)))
````

### Question 4.4 : Optimisation du résultat
- Pourriez-vous proposer une méthode qui optimise plus encore la précision du modèle sélectionné ?
- Pourriez-vous proposer une méthode qui permet de déterminer le meilleur modèle parmi une liste définie de modèles ?

````python

from sklearn.model_selection import GridSearchCV

# Optimisation des hyperparamètres du RandomForest par GridSearchCV
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}

grid_search = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)
grid_search.fit(X_train_scaled, y_train)

print(f"Meilleurs paramètres : {grid_search.best_params_}")
print(f"Meilleur score CV    : {grid_search.best_score_:.4f}")

best_model = grid_search.best_estimator_
print(f"\nScore Test avec le meilleur modèle : {best_model.score(X_test_scaled, y_test):.4f}")
````

````python

from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

# Comparaison de plusieurs modèles
candidates = {
    'RandomForest':   RandomForestClassifier(n_estimators=100, random_state=42),
    'DecisionTree':   DecisionTreeClassifier(random_state=42),
    'SVC':            SVC(random_state=42),
    'KNeighbors':     KNeighborsClassifier()
}

results = {}
for name, clf in candidates.items():
    clf.fit(X_train_scaled, y_train)
    results[name] = clf.score(X_test_scaled, y_test)

results_df = pd.DataFrame.from_dict(results, orient='index', columns=['Accuracy (Test)'])
results_df = results_df.sort_values('Accuracy (Test)', ascending=False)
print(results_df.round(4))
````

#### Amélioration : regroupement en 3 classes (low / medium / high)

Les scores assez bas s'expliquent par le fait que la cible **quality** comprend 7 classes très déséquilibrées (notes 3 à 9). 
Pour améliorer les performances, on regroupe les notes en **3 catégories** :

| Catégorie | Notes |
|-----------|-------|
| **low**    | ≤ 4   |
| **medium** | 5 – 6 |
| **high**   | ≥ 7   |

````python

# les 3 classes (low / medium / high)
def bin_quality(q):
    if q <= 4:
        return 'low'
    elif q <= 6:
        return 'medium'
    else:
        return 'high'

y_binned      = y.map(bin_quality)
y_train_b     = y_train.map(bin_quality)
y_test_b      = y_test.map(bin_quality)

print("Distribution des classes :")
print(y_binned.value_counts().sort_index())

# Comparaison des 4 modèles avec les 3 classes
candidates_b = {
    'RandomForest': RandomForestClassifier(n_estimators=100, random_state=42),
    'DecisionTree': DecisionTreeClassifier(random_state=42),
    'SVC':          SVC(random_state=42),
    'KNeighbors':   KNeighborsClassifier()
}

results_b = {}
for name, clf in candidates_b.items():
    clf.fit(X_train_scaled, y_train_b)
    results_b[name] = clf.score(X_test_scaled, y_test_b)

results_b_df = pd.DataFrame.from_dict(results_b, orient='index', columns=['Accuracy (Test) – 3 classes'])
results_b_df = results_b_df.sort_values('Accuracy (Test) – 3 classes', ascending=False)
print("\nComparaison des modèles (3 classes) :")
print(results_b_df.round(4))

# Meilleur modèle sur 3 classes + rapport détaillé
best_name_b = results_b_df.index[0]
best_clf_b  = candidates_b[best_name_b]
from sklearn.metrics import classification_report
print(f"\nRapport de classification – {best_name_b} :")
print(classification_report(y_test_b, best_clf_b.predict(X_test_scaled)))
````
